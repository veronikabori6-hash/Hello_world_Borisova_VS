```python
from datetime import datetime
reagent_name = input("Введите название нового реактива: ").strip()
while True:
    try:
        reagent_quantity = int(input("Введите его количество (целое число): "))
        if reagent_quantity < 0:
            print("Количество не может быть отрицательным. Попробуйте снова.")
            continue
        break
    except ValueError:
        print("Ошибка! Введите целое число.")
        current_date = datetime.now().strftime("%d.%m.%Y %H:%M")
report = f"Реактив {reagent_name} поступил на склад в количестве {reagent_quantity} ."
print("\n" + "=" * 50)
print(report)
print("=" * 50)

print(f"Отчет сохранен в файл inventory.txt (запись от {current_date})")
print("\nСодержимое файла inventory.txt:")
print("-" * 30)
try:
    with open("inventory.txt", "r", encoding="utf-8") as file:
        print(file.read())
except FileNotFoundError:
    print("Файл еще не существует или был только что создан.")
```
