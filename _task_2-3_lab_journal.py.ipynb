```python
from datetime import datetime

def wrap_text(text, max_width):
    """Разбивает текст на строки заданной максимальной ширины"""
    words = text.split()
    lines = []
    current_line = ""
    
    for word in words:
        if len(current_line) + len(word) + 1 <= max_width:
            if current_line:
                current_line += " " + word
            else:
                current_line = word
        else:
            if current_line:
                lines.append(current_line)
            current_line = word
    
    if current_line:
        lines.append(current_line)
    
    return lines

# Собираем данные от пользователя
print("=" * 60)
print("ЗАПОЛНЕНИЕ ЭЛЕКТРОННОГО ЛАБОРАТОРНОГО ЖУРНАЛА")
print("=" * 60)

researcher_name = input("Введите ФИО исследователя: ").strip()

# Автоматическая подстановка текущей даты, если пользователь ничего не ввел
date_input = input("Введите дату (Enter для текущей даты): ").strip()
if date_input:
    experiment_date = date_input
else:
    experiment_date = datetime.now().strftime("%d.%m.%Y")
    print(f"Использована текущая дата: {experiment_date}")

experiment_name = input("Введите название эксперимента: ").strip()
experiment_conclusion = input("Введите вывод по эксперименту: ").strip()

# Определяем ширину рамки
frame_width = 60

# Разбиваем вывод на строки
conclusion_lines = wrap_text(experiment_conclusion, frame_width - 6)  # -6 для учета отступов

# Создаем файл journal.txt и записываем данные в рамке
with open("journal.txt", "w", encoding="utf-8") as file:
    # Верхняя граница рамки
    file.write("╔" + "═" * (frame_width - 2) + "╗\n")
    
    # Заголовок
    title = "ЭЛЕКТРОННЫЙ ЛАБОРАТОРНЫЙ ЖУРНАЛ"
    file.write(f"║ {title:<{frame_width - 3}}║\n")
    
    # Разделитель
    file.write("╠" + "═" * (frame_width - 2) + "╣\n")
    
    # Данные исследователя
    file.write(f"║ ФИО исследователя : {researcher_name:<{frame_width - 23}}║\n")
    file.write(f"║ Дата             : {experiment_date:<{frame_width - 23}}║\n")
    file.write(f"║ Эксперимент      : {experiment_name:<{frame_width - 23}}║\n")
    
    # Разделитель
    file.write("╠" + "═" * (frame_width - 2) + "╣\n")
    
    # Заголовок для вывода
    file.write(f"║ ВЫВОД:\n")
    file.write(f"║" + " " * (frame_width - 2) + "║\n")
    
    # Записываем вывод построчно
    if conclusion_lines:
        for line in conclusion_lines:
            file.write(f"║ {line:<{frame_width - 4}} ║\n")
    else:
        file.write(f"║ {'Нет данных':<{frame_width - 4}} ║\n")
    
    # Нижняя граница рамки
    file.write("╚" + "═" * (frame_width - 2) + "╝\n")

print("\n" + "=" * 60)
print("✅ Файл 'journal.txt' успешно создан!")
print("=" * 60)

# Для наглядности выводим содержимое файла в консоль
print("\nСодержимое файла journal.txt:")
print("-" * frame_width)
with open("journal.txt", "r", encoding="utf-8") as file:
    print(file.read())
```
