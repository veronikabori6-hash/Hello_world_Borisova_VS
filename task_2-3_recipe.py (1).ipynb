```python
# task_2-3_recipe.py - Расширенная версия

# Запрашиваем у пользователя данные о питательной среде
medium_name = input("Введите название питательной среды: ").strip()

# Проверка ввода концентрации агара
while True:
    agar_input = input("Введите концентрацию агара (%): ").strip()
    try:
        agar_concentration = float(agar_input)
        if agar_concentration < 0 or agar_concentration > 100:
            print("Концентрация должна быть в диапазоне от 0 до 100. Попробуйте снова.")
            continue
        break
    except ValueError:
        print("Ошибка! Введите число (можно использовать десятичную точку).")

# Проверка ввода температуры стерилизации
while True:
    temp_input = input("Введите температуру стерилизации (°C): ").strip()
    try:
        sterilization_temp = float(temp_input)
        if sterilization_temp < 0 or sterilization_temp > 200:
            print("Температура должна быть в разумных пределах (0-200°C). Попробуйте снова.")
            continue
        break
    except ValueError:
        print("Ошибка! Введите число.")

# Добавляем текущую дату для информации
from datetime import datetime
current_date = datetime.now().strftime("%d.%m.%Y")

# Создаем файл recipe.txt и записываем данные с красивым форматированием
with open("recipe.txt", "w", encoding="utf-8") as file:
    # Записываем заголовок с обрамлением
    file.write("=" * 50 + "\n")
    file.write(f"РЕЦЕПТ ПИТАТЕЛЬНОЙ СРЕДЫ\n")
    file.write("=" * 50 + "\n\n")
    
    # Записываем название среды как основной заголовок
    file.write(f"Название: {medium_name}\n")
    file.write("─" * 30 + "\n\n")
    
    # Записываем параметры в виде структурированного списка
    file.write("ПАРАМЕТРЫ:\n")
    file.write(f"  • Концентрация агара . . . . . . . {agar_concentration} %\n")
    file.write(f"  • Температура стерилизации . . . . {sterilization_temp} °C\n\n")
    
    # Добавляем дату создания
    file.write(f"Дата создания рецепта: {current_date}\n")
    file.write("=" * 50 + "\n")

# Выводим сообщение об успешном завершении
print(f"\nФайл 'recipe.txt' успешно сформирован!")

# Для наглядности показываем содержимое созданного файла
print("\nСодержимое файла recipe.txt:")
print("-" * 30)
try:
    with open("recipe.txt", "r", encoding="utf-8") as file:
        print(file.read())
except FileNotFoundError:
    print("Файл не найден.")
```
