```python
#!/usr/bin/env python3
# -*- coding: utf-8 -*-

"""
Программа для перебора нескольких последовательностей ДНК с нумерацией позиций.
"""

# Список последовательностей ДНК
seq = ["ATATACGCGTA", "CTTCGGNGGA"]

print("=== Анализ последовательностей ДНК ===\n")

# Перебор каждой последовательности в списке
for i, sequence in enumerate(seq, 1):
    print(f"Последовательность {i} (длина: {len(sequence)}):")
    print(f"Целиком: {sequence}")
    
    print("Построчно (с нумерацией позиций):")
    # Перебор каждого нуклеотида в последовательности с индексом
    for position, nucleotide in enumerate(sequence, 1):
        print(f"Позиция {position}: {nucleotide}")
    
    print("-" * 30)  # Разделитель между последовательностями

print("\nЦикл выполнен")
```
