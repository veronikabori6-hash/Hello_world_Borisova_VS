```python

from datetime import datetime

while True:
    operator_name = input("Введите имя оператора: ").strip()
    if operator_name:
        break
    print("Имя оператора не может быть пустым. Попробуйте снова.")


while True:
    try:
        pressure_value = float(input("Введите текущее значение давления (Па): "))
        if pressure_value < 0:
            print("Давление не может быть отрицательным. Попробуйте снова.")
            continue
        break
    except ValueError:
        print("Ошибка! Введите число (можно использовать десятичную точку).")

current_datetime = datetime.now().strftime("%d.%m.%Y %H:%M:%S")


import os

file_exists = os.path.isfile("sensor_log.txt")

with open("sensor_log.txt", "a", encoding="utf-8") as file:
 
    if not file_exists:
        file.write("ДАТА И ВРЕМЯ\t\tОПЕРАТОР\tЗНАЧЕНИЕ (Па)\n")
        file.write("-" * 60 + "\n")
  
    file.write(f"{current_datetime}\t{operator_name}\t{pressure_value}\n")

print("\nДанные успешно сохранены в sensor_log.txt")

print("\nПоследние записи в файле sensor_log.txt:")
print("-" * 50)
try:
    with open("sensor_log.txt", "r", encoding="utf-8") as file:
        lines = file.readlines()
    
        for line in lines[-5:]:
            print(line.rstrip())
except FileNotFoundError:
    print("Файл не найден.")
```
