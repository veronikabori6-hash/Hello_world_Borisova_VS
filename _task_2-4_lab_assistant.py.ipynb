```python
from datetime import datetime

print("╔" + "═" * 58 + "╗")
print("║         ЛАБОРАТОРНЫЙ ПОМОЩНИК - ФИЗИОЛОГИЧЕСКИЙ РАСТВОР         ║")
print("║                   0.9% NaCl (изотонический)                     ║")
print("╚" + "═" * 58 + "╝")

# Запрашиваем данные с проверкой
while True:
    try:
        volume = float(input("▶ Введите необходимый объем раствора (мл): "))
        if volume <= 0:
            print("❌ Объем должен быть положительным числом. Попробуйте снова.")
            continue
        if volume > 5000:
            print("⚠️  Предупреждение: большой объем раствора!")
        break
    except ValueError:
        print("❌ Ошибка! Введите число (можно использовать десятичную точку).")

# Дополнительно: спрашиваем название эксперимента (опционально)
experiment_name = input("▶ Введите название эксперимента (или нажмите Enter): ").strip()
operator_name = input("▶ Введите имя оператора (или нажмите Enter): ").strip()

# Расчеты
# Концентрация 0.9% означает 0.9 г соли на 100 мл раствора
# Следовательно, масса соли (г) = объем (мл) * 0.009
salt_mass = volume * 0.009

# Объем воды (приблизительно равен общему объему)
water_volume = volume

# Точный расчет с учетом плотности раствора (для информации)
# Плотность 0.9% NaCl примерно 1.004 г/мл
solution_mass = volume * 1.004
water_mass = solution_mass - salt_mass
exact_water_volume = water_mass / 1.0  # плотность воды 1 г/мл

# Округление значений для отображения
volume_display = f"{volume:.2f}"
salt_display = f"{salt_mass:.2f}"
water_display = f"{water_volume:.2f}"
exact_water_display = f"{exact_water_volume:.2f}"

# Вывод в консоль с красивым форматированием
print("\n" + "┌" + "─" * 58 + "┐")
print("│{:^58}│".format("РЕЗУЛЬТАТ РАСЧЕТА"))
print("├" + "─" * 58 + "┤")

# Форматируем вывод с выравниванием
print("│ {:<30} {:>25} │".format("Общий объем раствора:", f"{volume_display} мл"))
print("│ {:<30} {:>25} │".format("Масса NaCl (0.9%):", f"{salt_display} г"))
print("│ {:<30} {:>25} │".format("Объем воды (приблизительно):", f"{water_display} мл"))
print("├" + "─" * 58 + "┤")
print("│ {:<30} {:>25} │".format("Точный объем воды (с учетом):", f"{exact_water_display} мл"))
print("│ {:<30} {:>25} │".format("Масса готового раствора:", f"{solution_mass:.2f} г"))
print("└" + "─" * 58 + "┘")

# Сохранение в файл
filename = "recipe.txt"

with open(filename, "w", encoding="utf-8") as file:
    # Заголовок
    file.write("=" * 50 + "\n")
    file.write("ОТЧЕТ ПО ПРИГОТОВЛЕНИЮ ФИЗИОЛОГИЧЕСКОГО РАСТВОРА\n")
    file.write("=" * 50 + "\n\n")
    
    # Мета-информация, если указана
    if experiment_name:
        file.write(f"Эксперимент: {experiment_name}\n")
    if operator_name:
        file.write(f"Оператор: {operator_name}\n")
    
    # Дата и время
    current_time = datetime.now().strftime("%d.%m.%Y %H:%M")
    file.write(f"Дата и время: {current_time}\n\n")
    
    # Разделитель
    file.write("-" * 30 + "\n\n")
    
    # Основные параметры
    file.write(f"Общий объем: [{volume_display}] мл\n")
    file.write(f"Масса соли:  [{salt_display}] г\n")
    file.write(f"Объем воды:  [{water_display}] мл\n\n")
    
    # Дополнительная информация
    file.write("-" * 30 + "\n")
    file.write("ПРИМЕЧАНИЕ:\n")
    file.write("• Для приготовления используйте дистиллированную воду\n")
    file.write("• Тщательно перемешайте до полного растворения соли\n")
    file.write(f"• Концентрация: 0.9% NaCl\n")
    file.write("=" * 50 + "\n")

print(f"\n✅ Рецепт успешно сохранен в файл '{filename}'")

# Показываем содержимое файла
print("\n" + "┌" + "─" * 58 + "┐")
print("│{:^58}│".format("СОДЕРЖИМОЕ ФАЙЛА recipe.txt"))
print("└" + "─" * 58 + "┘")
print()

with open(filename, "r", encoding="utf-8") as file:
    print(file.read())
```
