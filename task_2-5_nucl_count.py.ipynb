```python
#!/usr/bin/env python3
# -*- coding: utf-8 -*-

"""
Программа для анализа последовательности ДНК.
Подсчитывает количество каждого нуклеотида, общую длину и процентное содержание.
"""

def main():
    """Основная функция программы."""
    
    print("=== Анализ последовательности ДНК ===\n")
    
    dna_sequence = input("Введите последовательность ДНК: ").strip()
  
    if not dna_sequence:
        print("Ошибка: введена пустая строка!")
        return
  
    dna_upper = dna_sequence.upper()
    
    print(f"\nПоследовательность в верхнем регистре: {dna_upper}\n")
    

    count_a = dna_upper.count('A')
    count_t = dna_upper.count('T')
    count_g = dna_upper.count('G')
    count_c = dna_upper.count('C')
    
   
    total_length = len(dna_upper)
  
    print("Подсчёт нуклеотидов:")
    print(f"A: {count_a}")
    print(f"T: {count_t}")
    print(f"G: {count_g}")
    print(f"C: {count_c}")
    
    print(f"\nОбщая длина: {total_length} нуклеотидов\n")
    
    print("Процентное содержание:")
    if total_length > 0:
        print(f"A: {(count_a / total_length * 100):.1f}%")
        print(f"T: {(count_t / total_length * 100):.1f}%")
        print(f"G: {(count_g / total_length * 100):.1f}%")
        print(f"C: {(count_c / total_length * 100):.1f}%")
    else:
        print("Невозможно вычислить проценты для пустой последовательности")

if __name__ == "__main__":
    main()
```
